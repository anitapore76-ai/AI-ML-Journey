```python
import pandas as pd
serie=pd.Series([80,30,20,10,90,85,45,35],dtype="int8",name="marks")
print(serie)
print(serie.name)
print(serie.ndim)
print(serie.shape)
print(serie.index)
print(serie.values)
print(serie.dtype)
print(serie.size)
print(serie.iloc[0])
print(serie.iloc[0:3])
print(serie.iloc[-1])
print(serie[serie>50])
```

```python
#select all marks greater than 80
print(serie[serie>80])
print(serie.between(40,90))
print(serie.isin([40,90,35]))
print(serie.notna())
print(serie.where(serie>50,10))
print(serie.replace(10,20))
print(serie.sort_values())
print(serie.sum())
print(serie.mean())
print(serie.std())
print(serie.max())
print(serie.min())
```

```python
#find missing values
print(serie.isna())
print(serie.isnull())
print(serie.notna())
print(serie.dropna())
print(serie.fillna(0))
```
