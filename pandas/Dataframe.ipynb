```python
import numpy as np
import pandas as pd
data={
    'name':['tom','nick','krish','jack','kabir'],
    'age':[20,np.nan,21,np.nan,19],
    'marks':[80,81,np.nan,82,83],
    'city':['delhi','mumbai',None,'pune','goa']
}
df=pd.DataFrame(data)
print(df)
```

```python
df.isna().sum()
```

```python
df.dropna()
```

```python
df.dropna(axis=1)
```

```python
df.fillna(0)
```

```python
df.sort_values(by='age')
```

```python
df.sort_index()
```

```python
df.groupby('city')['marks'].mean()
```

```python
df.groupby('city')['marks'].sum()
```

```python
df.groupby('city')['marks'].count()
```
